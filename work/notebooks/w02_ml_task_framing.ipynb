## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

```python
'''
Model Type: Regression and Ranking

Explanation: We will build a regression model to predict the expected CTR of a URL based on its current visibility and metadata features.
Once we have the expected CTR, we will calculate an "Opportunity Score" by subtracting the actual CTR from the predicted CTR. 
Finally, we will rank the URLs by score. Pages with the largest positive gap between predicted and actual CTR represent our highest-priority targets for a content refresh.
This will satisfy the chosen lane from earlier notebook.
'''
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
'''
Target variable: ctr (Click-through-rate). This is a continuous numerical value the model will learn to predict.

Key features (predictors): While feature engineering will happen later, the model will rely on inputs such as impressions (visibility),
current ranking position, and any available metadata indicators (e.g., title length, word count) present in the database.

Derived output: opportunity_score = predicted_ctr - actual_ctr
'''
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
'''
We will use two metrics: Root Mean Squared Error (RMSE) and the top 10% of the URLs recommended by our model have historically high volume of impressions.
RMSE will be used to measure the average difference between the model's predicted CTR and actual CTR. 

$$RMSE = \sqrt{\frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2}$$
'''
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
'''
A fixed rule relies on hardcoded logic, such as arbitrarily deciding a baseline score is exactly 
0.40 * visibility_score + 0.30 * freshness_risk_score.

The pattern for identifying CTR opportunities is too messy for a simple if-statement because it requires dynamically adjusting expected clicks based on position tiers, historical volume, and engagement context simultaneously. 
A fixed rule relies on rigid, hardcoded weights that cannot easily capture non-linear relationships.

The starter model results clearly prove this: a fixed baseline rule achieved a Precision@50 of 0.240 (getting only about 12 of its top 50 recommendations right), while a learned ranking using a random forest achieved a Precision@50 of 0.740 (getting about 37 out of 50 right). 
This provides strong evidence that a machine learning model is vastly superior at processing complex search signals to prioritize action queues.
'''
```
