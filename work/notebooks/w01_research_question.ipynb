## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# 1. Load the starter dataset
# The path goes up two levels from work/notebooks/ to reach data/raw/
file_path = '../../data/raw/content_refresh_anonymized.csv'
df = pd.read_csv(file_path)

# 2. Inspect the columns to ensure we have 'ctr' (Click-Through Rate)
# If the column is named differently (e.g., 'CTR' or 'click_through_rate'), update the variable below.
ctr_column = 'ctr' 

# If the dataset only provides 'clicks' and 'impressions', uncomment the line below to create the CTR column:
# df['ctr'] = df['clicks'] / df['impressions']

if ctr_column in df.columns:
    # 3. Calculate the numbers for the framing narrative
    total_pages = len(df)
    avg_ctr = df[ctr_column].mean()
    
    # Calculate how many pages fall below that average
    pages_below_avg = len(df[df[ctr_column] < avg_ctr])
    pct_below_avg = (pages_below_avg / total_pages) * 100
    
    # 4. Print the results to drop into your markdown
    print(f"Total Pages: {total_pages:,}")
    print(f"Average CTR: {avg_ctr:.2%}")
    print(f"Pages Below Average CTR: {pct_below_avg:.1f}%")
else:
    print(f"Column '{ctr_column}' not found. Here are the available columns to adjust:")
    print(df.columns.tolist())
    display(df.head())
```
